```python
%sql
CREATE OR REPLACE TABLE tabela_dre_final AS
SELECT 
    COALESCE(v.data_real, d.data_real) AS data,
    COALESCE(v.qtd_pizzas, 0) AS qtd_pizzas,
    COALESCE(v.faturamento_diario, 0.0) AS faturamento,
    COALESCE(d.custo_diario, 0.0) AS despesas,
    ROUND(COALESCE(v.faturamento_diario, 0.0) - COALESCE(d.custo_diario, 0.0), 2) AS lucro_liquido
FROM tabela_gold_vendas_pizzaria v
FULL OUTER JOIN tabela_gold_despesas_pizzaria d
    ON v.data_real = d.data_real
ORDER BY data;
```
